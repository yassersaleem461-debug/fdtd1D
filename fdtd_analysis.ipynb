```python
import numpy as np
import matplotlib.pyplot as plt
x = np.loadtxt("RealspaceGrid.txt")
x = x/1e-6 # convert to micro meters
y = np.zeros_like(x)

fig, ax = plt.subplots(figsize=(10,2))

ax.scatter(x,y, marker="|", s=500)
ax.set_xlabel(r"$x$ ($\mu$m)")
ax.set_yticks([])
ax.set_xticks(np.arange(0,41,4))
ax.grid(axis="x")
x
```

```python
epsilon = np.loadtxt("Epsilon_x.txt")
epsilon_0= 8.8541878128e-12
epsilon=epsilon/epsilon_0
plt.plot(x,epsilon)
plt.xlabel(r"$x$ ($\mu$m)")
plt.ylabel(r"$\epsilon(x)$")
```

```python
from IPython.display import clear_output
E_xt=np.loadtxt("FullEfield.txt")
print(E_xt.shape)
print(x.shape)
for n in range(0, E_xt.shape[0],1):
    clear_output(wait=True)
    plt.plot(x, E_xt[n,:])
    plt.xlabel(r"$x$ ($\mu$m)")
    plt.ylabel(r"$E(x,t)$")
    plt.ylim(-0.5, 0.5)
    plt.show()
    plt.axvline(x=24,color="red",linestyle="--")
```
