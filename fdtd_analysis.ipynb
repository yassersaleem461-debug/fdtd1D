```python
import numpy as np
import matplotlib.pyplot as plt

x = np.loadtxt("RealspaceGrid.txt")
y = np.zeros_like(x)

fig, ax = plt.subplots(figsize=(10,2))

ax.scatter(x,y, marker="|", s=500)
ax.set_xlabel(r"$x$ ($\mu$m)")
ax.set_yticks([])
ax.set_xticks(np.arange(0,41,4))
ax.grid(axis="x")
x
```
